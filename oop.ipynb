```python
class Auto:
    def __init__(self, znacka, rychlost, cena):
        self.znacka = znacka
        self.rychlost = rychlost
        self.cena = cena

    def info(self):
        print(f"Auto značky {self.znacka} jede {self.rychlost} km/h a stojí {self.cena} Kč.")

    def zrychli(self, kolik):
        self.rychlost += kolik

    def zmen_cenu(self, nova_cena):
        self.cena = nova_cena


moje_auto = Auto("Škoda", 50, 300000)

moje_auto.info()

moje_auto.zrychli(20)
moje_auto.zmen_cenu(280000)

moje_auto.info()
```
